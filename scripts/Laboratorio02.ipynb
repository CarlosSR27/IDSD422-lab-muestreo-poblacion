## **PARTE B: Población vs. Muestra**

```python
# ---------------------------------------------------------------------------
# Función auxiliar de reporte
# ---------------------------------------------------------------------------
def reporte(nombre, muestra, parametro):
    """Imprime n, media estimada y composición por ciudad de una muestra."""
    if muestra is None:
        print(f"[{nombre}] pendiente de completar (TODO)")
        return
    n = len(muestra)
    media = muestra["ingresos_mensuales"].mean()
    comp = muestra["ciudad"].value_counts().to_dict()
    print(f"[{nombre}] n={n}  media={media:.2f} USD  "
          f"(parámetro {parametro:.2f}, diferencia {media - parametro:+.2f})")
    print(f"    composición por ciudad: {comp}")
```

```python
# ---------------------------------------------------------------------------
# 1. La población y su parámetro (N = 400)
# ---------------------------------------------------------------------------
poblacion = pd.read_csv("data/clientes_400.csv")
parametro = poblacion["ingresos_mensuales"].mean()

print(f"Población: {poblacion.shape[0]} clientes × {poblacion.shape[1]} columnas")
print(f"Parámetro poblacional — ingreso medio: {parametro:.2f} USD")
print("Ciudades en la población:", poblacion["ciudad"].value_counts().to_dict())
```

```python
# ---------------------------------------------------------------------------
# 2. Muestra por conveniencia: "los primeros que llegaron" — TODO 1
# ---------------------------------------------------------------------------
# TODO 1: toma como muestra las primeras 20 filas de la población (.head(20)).

muestra_conv = poblacion.head(20)
reporte("Conveniencia", muestra_conv, parametro)
```

```python
# ---------------------------------------------------------------------------
# 3.1 Muestra aleatoria simple (semilla 1) — TODO 2
# ---------------------------------------------------------------------------
# TODO 2: sortea 20 clientes con la misma probabilidad para todos (.sample(20, random_state=1)).

muestra_aas = poblacion.sample(20, random_state=1)
reporte("Aleatorio simple", muestra_aas, parametro)
```

```python
# ---------------------------------------------------------------------------
# 3.2 Variabilidad entre semillas 1 a 5 — TODO 3
# ---------------------------------------------------------------------------
# TODO 3: repite el sorteo con las semillas 1 a 5 y anota las 5 medias en el informe.

print("Variabilidad del Muestreo Aleatorio Simple (semillas 1 a 5):")
filas_semillas = []
for semilla in range(1, 6):
    m_sem = poblacion.sample(20, random_state=semilla)
    media_sem = m_sem["ingresos_mensuales"].mean()
    dif_sem = media_sem - parametro
    comp_sem = m_sem["ciudad"].value_counts().to_dict()
    print(f"  Semilla {semilla}: media = {media_sem:.2f} USD (diferencia {dif_sem:+.2f} USD) | {comp_sem}")
    filas_semillas.append({
        "Semilla": semilla,
        "Media (USD)": round(media_sem, 2),
        "Diferencia vs Parámetro": round(dif_sem, 2)
    })

df_semillas = pd.DataFrame(filas_semillas)
print("\nTabla resumen de las 5 semillas:")
print(df_semillas.to_string(index=False))
```

```python
# ---------------------------------------------------------------------------
# 4. Muestra estratificada: 5 % dentro de cada ciudad — TODO 4
# ---------------------------------------------------------------------------
# TODO 4: muestrea dentro de cada ciudad con afijación proporcional (5 % de cada una).

muestra_est = poblacion.groupby("ciudad").sample(frac=0.05, random_state=1)
reporte("Estratificado", muestra_est, parametro)
```

```python
# ---------------------------------------------------------------------------
# 5. Comparación final de los tres diseños
# ---------------------------------------------------------------------------
reporte("Conveniencia", muestra_conv, parametro)
reporte("Aleatorio simple", muestra_aas, parametro)
reporte("Estratificado", muestra_est, parametro)
print()

if muestra_conv is not None and muestra_aas is not None and muestra_est is not None:
    resumen = pd.DataFrame([
        {"diseño": "Conveniencia (head)", "n": len(muestra_conv),
         "media estimada": round(muestra_conv["ingresos_mensuales"].mean(), 2),
         "diferencia": round(muestra_conv["ingresos_mensuales"].mean() - parametro, 2)},
        {"diseño": "Aleatorio simple (sample)", "n": len(muestra_aas),
         "media estimada": round(muestra_aas["ingresos_mensuales"].mean(), 2),
         "diferencia": round(muestra_aas["ingresos_mensuales"].mean() - parametro, 2)},
        {"diseño": "Estratificado (groupby + frac)", "n": len(muestra_est),
         "media estimada": round(muestra_est["ingresos_mensuales"].mean(), 2),
         "diferencia": round(muestra_est["ingresos_mensuales"].mean() - parametro, 2)},
    ])
    print(f"Parámetro poblacional: {parametro:.2f} USD")
    print(resumen.to_string(index=False))
```
